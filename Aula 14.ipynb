**Objetivo:** Compreender, implementar e comparar os principais algoritmos de ordenação. Aprender a analisar suas complexidades de tempo (melhor, médio e pior caso) e entender quando cada algoritmo é mais adequado.

### **Parte 1: Introdução e Algoritmos Simples (Complexidade O(n²))**

**Ordenar** é o processo de arranjar os itens de uma lista em uma sequência específica, como ordem crescente ou decrescente. A eficiência de muitas aplicações depende de dados bem ordenados.

#### **1. Método da Bolha (Bubble Sort)**

Este é um dos algoritmos mais simples de entender. Ele funciona percorrendo a lista várias vezes, comparando cada par de elementos adjacentes e trocando-os de lugar se estiverem na ordem errada. A cada passagem, o maior elemento "borbulha" para o final da lista.

* **Analogia:** Imagine bolhas de ar subindo em um tanque de água. As bolhas maiores (os maiores números) sobem mais rápido para o topo (o final da lista).
* **Performance:** É muito lento para listas grandes, com complexidade de tempo de **O(n²)** na média. Seu melhor caso é O(n) se a lista já estiver ordenada e usarmos uma otimização para parar cedo.


```python
def bubble_sort(lista_desordenada):
    tamanho_lista = len(lista_desordenada)
    # Loop externo para controlar o número de passagens
    for i in range(tamanho_lista - 1):
        # Loop interno para as comparações e trocas
        for j in range(0, tamanho_lista - i - 1):
            # Compara elementos adjacentes
            if lista_desordenada[j] > lista_desordenada[j+1]:
                # Troca se estiverem na ordem errada
                lista_desordenada[j], lista_desordenada[j+1] = lista_desordenada[j+1], lista_desordenada[j]
    return lista_desordenada
```

```python
lista = [64, 34, 25, 12, 22, 11, 90, 10]
bubble_sort(lista)
```

#### **2. Inserção Direta (Insertion Sort)**

Este algoritmo constrói a lista final ordenada um item de cada vez. Ele funciona dividindo a lista em duas partes: uma sub-lista ordenada e uma sub-lista com os itens restantes a serem inseridos.

* **Analogia:** É como você organiza um baralho de cartas na mão. Você pega uma carta de cada vez e a insere na posição correta entre as cartas que já estão ordenadas na sua mão.
* **Performance:** Também tem complexidade de **O(n²)**, mas é muito eficiente para listas pequenas ou listas que já estão "quase ordenadas".


```python
def insertion_sort(lista_desordenada):
    # Começa do segundo elemento, pois o primeiro já forma uma sub-lista ordenada
    for i in range(1, len(lista_desordenada)):
        chave = lista_desordenada[i]
        j = i - 1
        # Move os elementos da parte ordenada que são maiores que a chave
        # para uma posição à frente de sua posição atual
        while j >= 0 and chave < lista_desordenada[j]:
            lista_desordenada[j + 1] = lista_desordenada[j]
            j -= 1
        # Insere a chave na sua posição correta
        lista_desordenada[j + 1] = chave
    return lista_desordenada
```

```python
lista = [64, 34, 25, 12, 22, 11, 90, 10]
print(lista)
insertion_sort(lista)
```

#### **3. Seleção Direta (Selection Sort)**

Este algoritmo também divide a lista em uma parte ordenada e outra desordenada. A cada passagem, ele encontra o menor elemento na porção desordenada e o troca com o primeiro elemento dessa porção.

  * **Analogia:** É como um processo de seleção manual. Em cada rodada, você "seleciona" o menor item que ainda não foi posicionado e o coloca em seu devido lugar no início da lista.
  * **Performance:** A complexidade é sempre **O(n²)**, independentemente da ordem inicial da lista. Sua vantagem é que ele faz o mínimo de trocas possíveis (uma por passagem).


```python
def selection_sort(lista_desordenada):
    tamanho_lista = len(lista_desordenada)
    for i in range(tamanho_lista):
        # Encontra o índice do menor elemento na parte desordenada
        idx_menor = i
        for j in range(i + 1, tamanho_lista):
            if lista_desordenada[j] < lista_desordenada[idx_menor]:
                idx_menor = j
        # Troca o menor elemento encontrado com o primeiro elemento da parte desordenada
        lista_desordenada[i], lista_desordenada[idx_menor] = lista_desordenada[idx_menor], lista_desordenada[i]
    return lista_desordenada
```

```python
lista = [64, 34, 25, 12, 22, 11, 90, 10]
print(lista)
selection_sort(lista)
```

### **Parte 2: Algoritmos Avançados (Complexidade O(n log n))**

Para listas grandes, os algoritmos O(n²) são inviáveis. Algoritmos mais avançados usam estratégias inteligentes para alcançar uma performance muito superior.

#### **1. Quicksort (Método da Troca e Partição)**

O Quicksort é um algoritmo de "divisão e conquista". Ele funciona escolhendo um elemento como **pivô** e **particionando** a lista ao redor dele, de modo que todos os elementos menores que o pivô fiquem à sua esquerda e todos os maiores, à sua direita. Depois, ele aplica o mesmo processo recursivamente às duas sub-listas.

  * **Performance:** Extremamente rápido na prática, com complexidade média de **O(n log n)**. No entanto, seu pior caso é **O(n²)**, que pode ocorrer se o pivô escolhido for sempre o menor ou o maior elemento.


```python
def partition(lista, inicio, fim):
    pivo = lista[inicio]
    ponteiro_esq = inicio + 1
    ponteiro_dir = fim
    
    done = False
    while not done:
        while ponteiro_esq <= ponteiro_dir and lista[ponteiro_esq] <= pivo:
            ponteiro_esq += 1
        while lista[ponteiro_dir] >= pivo and ponteiro_dir >= ponteiro_esq:
            ponteiro_dir -= 1
        if ponteiro_dir < ponteiro_esq:
            done = True
        else:
            lista[ponteiro_esq], lista[ponteiro_dir] = lista[ponteiro_dir], lista[ponteiro_esq]
            
    lista[inicio], lista[ponteiro_dir] = lista[ponteiro_dir], lista[inicio]
    return ponteiro_dir

def quick_sort_recursivo(lista, inicio, fim):
    if inicio < fim:
        ponto_divisao = partition(lista, inicio, fim)
        quick_sort_recursivo(lista, inicio, ponto_divisao - 1)
        quick_sort_recursivo(lista, ponto_divisao + 1, fim)

def quick_sort(lista):
    quick_sort_recursivo(lista, 0, len(lista) - 1)
    return lista
```

```python
lista = [64, 34, 25, 12, 22, 11, 90, 10]
print(lista)
quick_sort(lista)
```

#### **2. Timsort (O Algoritmo Padrão do Python)**

O Timsort é um algoritmo híbrido, combinando o melhor do **Insertion Sort** e do **Merge Sort**. É o algoritmo de ordenação padrão usado no Python.

  * **Como Funciona:**
    1.  Ele divide a lista em pequenos blocos (chamados "runs"), geralmente de 32 ou 64 elementos.
    2.  Usa o **Insertion Sort** para ordenar cada um desses pequenos blocos de forma muito rápida.
    3.  Usa o processo de **Merge** (do Merge Sort) para combinar esses blocos ordenados de forma eficiente até que a lista inteira esteja ordenada.
  * **Vantagem:** É altamente otimizado para dados do mundo real, que frequentemente contêm sequências já ordenadas (os "natural runs").
  * **Performance:** Excelente performance com complexidade de **O(n log n)** no pior caso.

### **Parte 3: Comparação e Conclusão (30 minutos)**

A escolha do algoritmo de ordenação correto depende do cenário.

#### **Tabela Comparativa de Complexidades**

| Algoritmo | Melhor Caso | Caso Médio | Pior Caso |
| :--- | :--- | :--- | :--- |
| **Bubble Sort** | O(n) | O(n²) | O(n²) |
| **Insertion Sort** | O(n) | O(n²) | O(n²) |
| **Selection Sort** | O(n²) | O(n²) | O(n²) |
| **Quicksort** | O(n log n) | O(n log n)| O(n²) |
| **Timsort** | O(n) | O(n log n)| O(n log n)|
| **Heap Sort** | O(n log n) | O(n log n)| O(n log n)|
| **Merge Sort**| O(n log n) | O(n log n)| O(n log n)|

*Da Tabela 11.1 e de outros capítulos do livro.*

#### **Quando Usar Cada Um?**

  * **Insertion Sort:** Para listas pequenas ou quase ordenadas.
  * **Selection Sort:** Se o custo de trocas de elementos for muito alto.
  * **Quicksort:** A escolha geral para performance, se o risco do pior caso O(n²) for aceitável.
  * **Merge Sort / Heap Sort:** Quando uma performance **garantida** de O(n log n) é necessária.
  * **Timsort:** Em Python, quase sempre a melhor escolha, pois já é a implementação padrão do método `.sort()` e da função `sorted()`.

```python
lista = [64, 34, 25, 12, 22, 11, 90, 10]
print(lista)
sorted_lista = sorted(lista)
print(sorted_lista)
```
