# Aula 13: Árvores Balanceadas (AVL e Vermelho-Preto)

**Objetivo:** Entender por que o balanceamento é crucial para o desempenho de árvores de busca. Aprender os conceitos, regras e mecanismos fundamentais por trás de duas das mais importantes árvores auto-balanceáveis: a Árvore AVL e a Árvore Vermelho-Preto.

---

### **Parte 1: O Problema do Desbalanceamento**

Na aula anterior, vimos a Árvore de Busca Binária (BST). Ela é fantástica para buscas, com uma performance média de **O(log n)**. Mas ela tem um "calcanhar de Aquiles": seu desempenho depende totalmente de sua **forma**.

#### **Cenário Ideal vs. Pior Cenário**

Imagine que queremos inserir os números `[50, 30, 70, 20, 40, 60, 80]` em uma BST. O resultado seria uma árvore "cheia" e balanceada:

    50
   /  \
 30    70
 / \   / \
20 40 60  80


Nesta árvore, para encontrar qualquer elemento, o caminho é curto. A performance é **O(log n)**.

Agora, imagine que inserimos os números já ordenados: `[10, 20, 30, 40, 50]`. O resultado seria:


10

20

30

40

50

`
Esta árvore se tornou uma **"lista encadeada disfarçada"**. Para encontrar o número 50, precisamos percorrer todos os nós. A performance caiu para **O(n)**, o mesmo que uma busca linear simples!

**O Problema Central:** Uma BST comum não tem nenhum mecanismo para se "consertar". Se os dados chegam de uma forma ruim, a árvore fica "torta" (desbalanceada) e perde toda a sua eficiência.

**A Solução:** Precisamos de árvores mais inteligentes, que se **auto-balanceiam** durante as operações de inserção e deleção para nunca ficarem "tortas".

### **Parte 2: Árvore AVL - O Equilíbrio Estrito (90 minutos)**

A Árvore AVL foi a primeira árvore de busca binária auto-balanceável a ser inventada. Pense nela como a "perfeccionista" das árvores.

#### **A Regra da AVL: Equilíbrio de Altura**

A regra da AVL é simples e estrita:
> Para qualquer nó na árvore, a **altura** de sua subárvore esquerda e a de sua subárvore direita **não podem diferir em mais de 1**.

Para garantir essa regra, a árvore AVL armazena em cada nó um **Fator de Balanceamento (FB)**.

`Fator de Balanceamento = altura(subárvore_direita) - altura(subárvore_esquerda)`

Em uma árvore AVL válida, o FB de qualquer nó só pode ser **-1, 0 ou 1**.

#### **O Mecanismo de Correção: Rotações**

O que acontece se, após uma inserção ou deleção, o FB de um nó se torna **-2** ou **+2**? A árvore ficou desbalanceada! Para consertá-la, a AVL executa uma operação chamada **Rotação**.

Uma rotação é uma reorganização local de nós que muda a estrutura da árvore para restaurar o equilíbrio, mas ==preserva== a propriedade de busca binária.

**Analogia:** Pense em um equilibrista em uma corda bamba. Se ele se inclina demais para um lado (FB = -2 ou +2), ele move sua vara (os nós da árvore) para se reequilibrar.

Existem 4 tipos de desbalanceamento que exigem rotações:

1.  **Desbalanceamento Esquerda-Esquerda (FB = -2):** A árvore está "pesada" para a esquerda. Corrigido com uma **Rotação Simples à Direita**.

    
      (A) -2        -> Rotação Direita ->         (B) 0
     /                                           /   \
    (B) -1                                     (C) 0 (A) 0
   /
  (C) 0
    

2.  **Desbalanceamento Direita-Direita (FB = +2):** A árvore está "pesada" para a direita. Corrigido com uma **Rotação Simples à Esquerda**.

    
  (A) +2        -> Rotação Esquerda ->         (B) 0
     \                                        /   \
      (B) +1                                (A) 0 (C) 0
         \
          (C) 0
    

3.  **Desbalanceamento Esquerda-Direita (FB = -2):** Causa um "joelho" na subárvore esquerda. Corrigido com uma **Rotação Dupla** (primeiro uma rotação à esquerda no filho, depois uma à direita no avô).

   (A) -2        Passo 1: Rotação        (A) -2        Passo 2: Rotação         (C) 0
  /              em B (Esquerda)         /              em A (Direita)         /   \
 (B) +1          ------------>         (C) -1          ------------->        (B) 0 (A) 0
    \                                  /
     (C) 0                           (B) 0

4.  **Desbalanceamento Direita-Esquerda (FB = +2):** Causa um "joelho" na subárvore direita. Corrigido com uma **Rotação Dupla**.
(A) +2         Passo 1: Rotação     (A) +2        Passo 2: Rotação        (C) 0
  \          em B (Direita)           \           em A (Esquerda)        /   \
  (B) -1       ------------\>         (C) +1        -------------\>    (A) 0 (B) 0
  /                                     \
(C) 0                                   (B) 0

**Conclusão sobre AVL:**
* **Prós:** Por ser estritamente balanceada, é extremamente rápida para buscas.
* **Contras:** As regras rígidas podem exigir muitas rotações durante inserções e deleções, tornando essas operações um pouco mais lentas.

*(Nota: A implementação completa de uma Árvore AVL é complexa e está fora do escopo de uma aula introdutória. O importante é entender o conceito do Fator de Balanceamento e das Rotações.)*


### **Parte 3: Árvore Vermelho-Preto - O Equilíbrio Pragmático**

A Árvore Vermelho-Preto (Red-Black Tree - RBT) é a árvore auto-balanceável mais comum na prática. Pense nela como a "pragmática": ela não busca a perfeição como a AVL, mas garante que a árvore seja "boa o suficiente" para manter a performance.

Em vez de uma regra de altura, a RBT usa um conjunto de **5 regras baseadas em cores** (Vermelho e Preto) para cada nó.

#### **As 5 Regras da Árvore Vermelho-Preto**

1.  **Regra da Cor:** Todo nó é **Vermelho** ou **Preto**.
2.  **Regra da Raiz:** A raiz da árvore é sempre **Preta**.
3.  **Regra da Folha:** Todas as folhas (nós `NIL` ou `None`) são consideradas **Pretas**.
4.  **Regra do Nó Vermelho:** Se um nó é **Vermelho**, então ambos os seus filhos devem ser **Pretos**. (Isso impede que dois nós vermelhos fiquem um seguido do outro num mesmo caminho).
5.  **Regra da Profundidade Preta:** Todo caminho simples de um nó até qualquer uma de suas folhas descendentes contém o **mesmo número de nós Pretos**.

Essas regras, especialmente a 4 e a 5, garantem que o caminho mais longo da raiz a uma folha não seja mais do que o dobro do caminho mais curto. Isso mantém a árvore aproximadamente balanceada e garante a performance **O(log n)**.

#### **O Papel das Rotações e Recolorações**
Assim como na AVL, as **rotações** são usadas para ajustar a estrutura da RBT. A diferença é que elas são acionadas por violações das 5 regras, não por um fator de balanceamento. Frequentemente, uma rotação é acompanhada por uma **recoloração** dos nós para garantir que todas as regras voltem a ser satisfeitas.

Por exemplo, imagine que uma inserção criou dois nós vermelhos seguidos (pai `P` e filho `F`), violando a Regra 4:
```

(Avo - Preto)

(Pai - Vermelho)

(Filho - Vermelho) \<-- VIOLAÇÃO\!

```
Para corrigir isso, a árvore pode executar uma **Rotação à Esquerda** no Avô, e em seguida uma **recoloração**, trocando as cores do Avô e do Pai, restaurando o equilíbrio e as regras.
A complexidade das regras torna a implementação um desafio, mas garante uma performance O(log n) com menos rotações do que a AVL em média.

O balanceamento é mantido através de **rotações** e **recolorações** dos nós após inserções e deleções.

#### **Comparação: AVL vs. Vermelho-Preto**

| Característica | Árvore AVL | Árvore Vermelho-Preto |
| :--- | :--- | :--- |
| **Balanceamento** | Estrito. Alturas diferem no máximo em 1. | Menos estrito. O caminho mais longo é no máximo 2x o mais curto. |
| **Busca** | Mais rápida, pois a árvore é mais compacta. | Rápida, mas pode ser ligeiramente mais lenta que a AVL. |
| **Inserção/Deleção**| Mais lentas, pois podem exigir múltiplas rotações. | Mais rápidas, pois exigem menos rotações (no máximo 2) e usam recolorações. |
| **Uso Prático** | Ideal para sistemas com muito mais leituras do que escritas. | Ideal para sistemas com um equilíbrio entre leituras e escritas. É a escolha padrão em muitas bibliotecas de programação (ex: `map` do C++, `TreeMap` do Java). |
